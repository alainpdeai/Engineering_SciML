## 演示页代码实验台
下面单元格直接抽取自演示网页的可运行代码（`cell-src`），按顺序执行即可复现网页上的交互结果。

```python

import numpy as np
import matplotlib.pyplot as plt

def square_wave_series(x, t, L=1.0, alpha=1.0, N=51):
    # 目标：f(x)=1 on (0,L/2)，f(x)=-1 on (L/2,L)
    # 正确系数只在 n≡2 (mod 4) 时非零：b_n=8/(n*pi)（见 ⑪ 页卡 1）
    out = np.zeros_like(x, dtype=float)
    for n in range(1, N + 1):
        if n % 4 == 2:
            b = 8.0 / (n * np.pi)
            out += b * np.sin(n * np.pi * x / L) * np.exp(-alpha * (n * np.pi / L)**2 * t)
    return out

x = np.linspace(0, 1, 2001)
away = (np.abs(x - 0.5) > 0.03) & (x > 0.03) & (x < 0.97)
target = np.where(x < 0.5, 1.0, -1.0)

print("t=0（纯 Gibbs，无衰减）：")
for N in (3, 11, 51):
    u = square_wave_series(x, t=0.0, N=N)
    print(f"  N={N:2d}  跳跃/边界外最大误差={np.max(np.abs(u[away]-target[away])):.4f}")

print("N=51，不同 t（看高频衰减）：")
for t in (0.0, 0.001, 0.01, 0.05):
    u = square_wave_series(x, t=t, N=51)
    print(f"  t={t:.3f}  峰值={np.max(np.abs(u)):.4f}")

plt.figure(figsize=(6.2, 3.2))
for N, ls in zip((3, 11, 51), ["--", "-.", "-"]):
    plt.plot(x, square_wave_series(x, 0.0, N=N), ls, label=f"N={N}")
plt.plot(x, target, color="gray", lw=1, label="target")
plt.xlabel("x"); plt.ylabel("u(x, 0)")
plt.title("Gibbs overshoot shrinks in width, not in peak height")
plt.legend(); plt.show()
  
```

```python

import numpy as np

L, c, N = 1.0, 1.0, 40

def initial_shape(x):
    x0 = 0.3 * L
    return np.where(x < x0, x / x0, (L - x) / (L - x0))

trapz = getattr(np, "trapezoid", getattr(np, "trapz", None))

def a_coeffs(N):
    n = np.arange(1, N + 1)
    x = np.linspace(0, L, 4001)
    f = initial_shape(x)
    return np.array([2.0 / L * trapz(f * np.sin(k * np.pi * x / L), x) for k in n])

a = a_coeffs(N)
omega = np.arange(1, N + 1) * np.pi * c / L

def u_field(x, t):
    n = np.arange(1, N + 1)
    return (np.sin(np.outer(x, n) * np.pi / L) * np.cos(omega * t)) @ a

def energy(t, dt=1e-5, nx=4001):
    x = np.linspace(0, L, nx)
    ut = (u_field(x, t + dt) - u_field(x, t)) / dt
    ux = np.gradient(u_field(x, t), x)
    return 0.5 * trapz(ut**2 + c**2 * ux**2, x)

for t in (0.0, 0.3, 0.7, 1.3, 2.0):
    print(f"t={t:.2f}  E={energy(t):.6f}")
  
```
